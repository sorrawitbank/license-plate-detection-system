```python
from ultralytics import YOLO
```

```python
dataset_v11_path = "../data/license-plate.v1i.yolo11/data.yaml"
dataset_v26_path = "../data/license-plate.v1i.yolo26/data.yaml"
```

# Evaluate

## Model YOLO11n

```python
model_11 = YOLO("../runs/detect/license-plate-yolo11n-1/weights/best.pt")
```

```python
metrics_11 = model_11.val(
    data=dataset_v11_path,
    plots=True,
    save_json=True,
    split="test",
    conf=0.5,
    name="license-plate-yolo11n-test",
)
```

```python
print(f"  - Precision (P): {metrics_11.box.p[0]:.3f}")
print(f"  - Recall (R): {metrics_11.box.r[0]:.3f}")
print(f"  - mAP50: {metrics_11.box.map50:.3f}")
print(f"  - mAP50-95: {metrics_11.box.map:.3f}")
print(f"  - Speed (preprocess): {metrics_11.speed['preprocess']:.2f}ms")
print(f"  - Speed (inference): {metrics_11.speed['inference']:.2f}ms")
print(f"  - Speed (postprocess): {metrics_11.speed['postprocess']:.2f}ms")
```

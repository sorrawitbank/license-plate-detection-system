```python
from ultralytics import YOLO
```

```python
dataset_v11_path = "../data/license-plate.v1i.yolo11/data.yaml"
dataset_v26_path = "../data/license-plate.v1i.yolo26/data.yaml"
```

# Training

## Model YOLO11n

```python
model_11 = YOLO("yolo11n.pt")
```

```python
model_11.train(
    data=dataset_v11_path,
    epochs=100,
    imgsz=640,
    batch=16,
    patience=50,
    seed=42,
    name="license-plate-yolo11n-1",
)
```

## Model YOLO26n

```python
model_26 = YOLO("yolo26n.pt")
```

```python
model_26.train(
    data=dataset_v26_path,
    epochs=100,
    imgsz=640,
    batch=16,
    patience=50,
    seed=42,
    name="license-plate-yolo26n-1",
)
```

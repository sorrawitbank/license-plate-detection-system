# Final Training and Evaluation with YOLO11n

```python
from ultralytics import YOLO
```

## Training

```python
dataset_v11_path = "../data/license-plate.v1i.yolo11/data.yaml"
```

```python
model_11 = YOLO("yolo11n.pt")
```

```python
model_11.train(
    data=dataset_v11_path,
    epochs=300,
    imgsz=640,
    batch=16,
    patience=50,
    seed=42,
    name="license-plate-yolo11n-final",
)
```

## Evaluation

```python
model_final = YOLO("../runs/detect/license-plate-yolo11n-final/weights/best.pt")
```

```python
metrics_final = model_final.val(
    data=dataset_v11_path,
    plots=True,
    save_json=True,
    split="test",
    conf=0.5,
    name="license-plate-yolo11n-final-test",
)
```

```python
print(f"  - Precision (P): {metrics_final.box.p[0]:.3f}")
print(f"  - Recall (R): {metrics_final.box.r[0]:.3f}")
print(f"  - mAP50: {metrics_final.box.map50:.3f}")
print(f"  - mAP50-95: {metrics_final.box.map:.3f}")
print(f"  - Speed (preprocess): {metrics_final.speed['preprocess']:.2f}ms")
print(f"  - Speed (inference): {metrics_final.speed['inference']:.2f}ms")
print(f"  - Speed (postprocess): {metrics_final.speed['postprocess']:.2f}ms")
```
